```python
# Day 7: Prompting vs Fine-Tuning
# PortfolioAI request classifier

training_examples = [
    ("Add my Python chatbot project", "project"),
    ("Include my RAG application", "project"),
    ("Show my GitHub projects", "project"),
    ("Update my technical skills", "skills"),
    ("Add Python and SQL to my profile", "skills"),
    ("List my machine learning skills", "skills"),
    ("Update my BTech degree details", "education"),
    ("Add my college information", "education"),
    ("Show my graduation year", "education"),
    ("Add my course certificates", "certifications"),
    ("Include my internship certificate", "certifications"),
    ("Display my completed AI courses", "certifications"),
]

print("Training examples:", len(training_examples))

for text, label in training_examples[:4]:
    print(f"Request: {text} -> Label: {label}")
```

```python
# Rule-based baseline representing a simple
# zero-shot-style classification strategy.

def classify_baseline(request):
    text = request.lower()

    if any(word in text for word in [
        "certificate", "certification", "course"
    ]):
        return "certifications"

    if any(word in text for word in [
        "degree", "college", "graduation", "education"
    ]):
        return "education"

    if any(word in text for word in [
        "skill", "python", "sql", "machine learning"
    ]):
        return "skills"

    if any(word in text for word in [
        "project", "github", "rag", "chatbot"
    ]):
        return "project"

    return "general"


test_requests = [
    "Add my RAG chatbot project",
    "Update my degree details",
    "Show my machine learning skills",
    "Include my AI course certificate",
    "Help me make my portfolio better",
]

for request in test_requests:
    prediction = classify_baseline(request)
    print(f"{request} -> {prediction}")
```

```python
# Example bank for few-shot-inspired classification.

few_shot_examples = [
    ("showcase my academic program", "education"),
    ("display my completed degree", "education"),
    ("publish my software project", "project"),
    ("show my coding abilities", "skills"),
    ("display my course completion proof", "certifications"),
]


def classify_with_examples(request):
    text = request.lower()

    # First, use the original baseline.
    baseline = classify_baseline(request)

    if baseline != "general":
        return baseline

    # For unseen requests, compare against example phrases.
    request_words = set(text.split())

    best_label = "general"
    best_score = 0

    for example_text, label in few_shot_examples:
        example_words = set(example_text.split())

        score = len(
            request_words.intersection(example_words)
        )

        if score > best_score:
            best_score = score
            best_label = label

    return best_label


new_requests = [
    "Please showcase my completed academic program",
    "Publish my software project",
    "Display my course completion proof",
    "Help me make my portfolio better",
]

for request in new_requests:
    print(
        request,
        "->",
        classify_with_examples(request)
    )
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline


# Training data
X_train = [text for text, label in training_examples]
y_train = [label for text, label in training_examples]


# Build a text classification pipeline
model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            lowercase=True,
            ngram_range=(1, 2)
        )
    ),
    (
        "classifier",
        LogisticRegression(
            max_iter=1000,
            random_state=42
        )
    )
])


# Train the model
model.fit(X_train, y_train)


# Evaluate on unseen examples
test_data = [
    ("Add my new software project", "project"),
    ("Show my technical skills", "skills"),
    ("Update my college degree", "education"),
    ("Display my course certificate", "certifications"),
]

X_test = [text for text, label in test_data]
y_test = [label for text, label in test_data]

predictions = model.predict(X_test)

print("\nTrained classifier results:")

for request, expected, predicted in zip(
    X_test, y_test, predictions
):
    print(f"Request: {request}")
    print(f"Expected: {expected}")
    print(f"Predicted: {predicted}")
    print()

accuracy = sum(
    predicted == expected
    for predicted, expected in zip(y_test, predictions)
) / len(y_test)

print(f"Test accuracy: {accuracy:.2f}")
```

```python
# Compare baseline, example-based rules, and trained classifier.

comparison_requests = [
    "Add my RAG chatbot project",
    "Update my degree details",
    "Show my machine learning skills",
    "Include my AI course certificate",
    "Please showcase my completed academic program",
    "Display my course completion proof",
]

print("\nCOMPARISON")

for request in comparison_requests:
    baseline = classify_baseline(request)
    example_based = classify_with_examples(request)
    trained = model.predict([request])[0]

    print("\nRequest:", request)
    print("Baseline:", baseline)
    print("Example-based:", example_based)
    print("Trained classifier:", trained)
```
